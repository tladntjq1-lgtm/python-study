## 2. 작업 계획

### 전체 작업 순서
1. records, columns 데이터가 어떤 구조인지 확인한다.
2. 가입일 최댓값/최솟값, 나이 합계, 카운트의 초기값과 소문자 컬럼명을 담을 빈 리스트를 준비한다.
3. records를 한 명씩 반복하며 나이를 합계 변수에 더하고 카운트를 늘린다.
4. 같은 반복 안에서 현재 고객의 가입일을 최댓값/최솟값 변수와 비교하며 값을 갱신한다.
5. records 반복이 끝나면 합계와 카운트를 이용해 평균을 계산한다.
6. columns를 별도의 반복문으로 하나씩 확인한다.

8. 최종 결과(가입일 최댓값/최솟값, 나이 합계/평균, 소문자 컬럼명)를 출력한다.

### 반복해야 하는 것은 무엇인가?
records와 columns 둘 다 반복해야 한다. records는 가입일 최댓값/최솟값, 나이 합계/평균을 구하기 위해 반복하고, columns는 컬럼명을 소문자로 바꾸기 위해 반복한다. 각각 원하는 결과가 다르기 때문에 따로 반복해야 한다.

### 반복하면서 계속 기억해야 하는 값은 무엇인가?
- 가입일 최댓값 (가장 최근 가입일)
- 가입일 최솟값 (가장 오래된 가입일)
- 나이 합계
- 카운트(지금까지 확인한 사람 수)

### 조건으로 비교해야 하는 값은 무엇인가?
현재 고객의 가입일을 지금까지 기억해둔 최댓값(가장 최신)과 비교해서 누가 더 최신인지 판단하고, 지금까지 기억해둔 최솟값(가장 오래됨)과 비교해서 누가 더 오래됐는지 판단한다.

### 반복이 모두 끝난 뒤 계산해야 하는 것은 무엇인가?
평균. 합계는 반복 중에 계속 누적하는 값이고, 반복이 끝난 뒤에는 그 합계와 카운트를 이용해 평균만 새로 계산하면 된다.

- 어떤 경우에 값을 바꿀 것인가: 현재 가입일이 지금까지의 가장 큰 값보다 크면 newest_date를 그 값으로 바꾸고, 가장 작은 값보다 작으면 oldest_date를 그 값으로 바꾼다. 둘 다 아니면 그대로 두고 다음 고객으로 넘어간다.
- 아직 헷갈리는 부분: 없음


````python
Oldest_date = records[0]["Signup_Date"]
Newest_date = records[0]["Signup_Date"]

for rec in records:
    oddt = rec["Signup_Date"]
    nedt = rec["Signup_Date"]
    if oddt < Oldest_date:
        Oldest_date = oddt
    if nedt > Newest_date:
        Newest_date = nedt

print(Oldest_date, Newest_date)
````

### 5-3. 나이 합계/평균 구현 전 생각
- 합계를 저장할 변수: age_sum
- 처음 값은 무엇으로 둘 것인가: 0
- 고객 한 명을 확인할 때 무엇을 더할 것인가: rec["Age"]
- 사람 수는 어떻게 셀 것인가: count 변수를 만들어서 한 명 확인할 때마다 1씩 늘린다
- 평균은 언제 계산할 것인가: records 반복이 모두 끝난 뒤, age_sum / count로 계산
- 아직 헷갈리는 부분: 없음


````python
age_sum = 0
count = 0

for rec in records:
    age_sum = age_sum + rec["Age"]
    count = count + 1

age_avg = age_sum / count

print(age_sum, age_avg)
````
