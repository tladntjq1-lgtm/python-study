2. 가장 중요한 규칙 — 문제 해결에는 pandas 사용 금지

이번 퀴즈에서는 데이터를 준비하는 첫 번째 셀까지만 pandas를 사용합니다. 아래 준비 코드는 그대로 실행합니다.

이 셀에서는 CSV를 DataFrame으로 읽고, 컬럼명을 일부러 대문자가 포함된 형태로 바꾼 뒤, DataFrame의 내용을 Python 기본 자료구조로 꺼내 줍니다.

python
import pandas as pd

url = "https://raw.githubusercontent.com/GilbertMoon/llm-data-analysis-course/main/data/raw/customers.csv"

df = pd.read_csv(url)

df.columns = [
    "Customer_ID",
    "Name",
    "Gender",
    "Age",
    "City",
    "Signup_Date"
]

records = df.to_dict(orient="records")
columns = list(df.columns)

print(df.head())

위 셀은 데이터 준비용으로 제공되는 코드입니다. 이 셀 아래부터 학생이 작성하는 문제 해결 코드에서는 pandas 기능을 사용하지 않습니다.

이후에는 다음 두 변수만 사용해서 문제를 해결합니다.

records → 전체 고객 데이터가 들어 있는 List
columns → 전체 컬럼명이 들어 있는 List

records 안의 고객 한 명은 Dictionary입니다. 즉 Chapter 01에서 계속 연습했던 구조와 같습니다.

List
 ↓
Dictionary 여러 개
 ↓
for로 하나씩 확인
 ↓
필요한 값 꺼내기
 ↓
조건 판단 / 값 누적
 ↓
결과 만들기
3. 사용 금지 기능

이번 퀴즈는 Python의 기본 반복과 조건을 직접 사용하는 것이 목적입니다. 따라서 학생이 작성하는 문제 해결 코드에서는 다음 기능을 사용하지 않습니다.

✗ df["Signup_Date"].min()

✗ Series 또는 DataFrame의 집계 기능
✗ max()
✗ min()
✗ sum()
✗ statistics.mean()

AI에게 다음을 요청해서도 안 됩니다.

✗ 정답 코드 작성
