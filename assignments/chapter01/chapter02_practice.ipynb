````python
students= [
    {"name":"민수","score":95, "attendance":90},
    {"name":"지영","score":85, "attendance":95},
    {"name":"현우","score":72, "attendance":65},
    {"name":"수진","score":78, "attendance":88},
]
for student in students:
    name=student["name"]
    score=student["score"]
    attendance=student["attendance"]
    if score<60 or attendance<70:
        grade="F"
    elif score>=90 and attendance>=80:
        grade="A"
    elif score>=80 and attendance>=80:
        grade="B"
    elif score>=70:
        grade="C"
    else:
        grade="D"
    print(f"{name}님의 점수는 {score}점 출석률은{attendance}% 로 {grade}등급입니다")
````
