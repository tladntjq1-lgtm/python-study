```python
students=[
    {"name": "민수","score":85},
    {"name": "지영","score":72},
    {"name": "현우","score":91},
]
for student in students:
    name=student["name"]
    score=student["score"]
    if score>=80:
        result="합격"
    else:
        result="불합격"
    print(name, score, result)
```
